# questions

```python
#1. what is the average price and average mileage for convertible cars

df[df['body_style'] == 'convertible'][['price','mileage']].mean()
```

```python
 # 2. Sort cars by company and then price in descending order
df.sort_values(by = ['company','price'],ascending=False)
```

```python
# 3. Save data for 'Toyota' separately
toyota = df[df['company'] == 'toyota']
toyota.to_csv('toyota.csv')
```

```python
# 4 which company is producing most expensive car?
df[['company','price']].sort_values(by='price',ascending = False).head(1)
```

```python
# 5. Name the cars on basis of maximum horsepower
df[['body_style','horsepower']].sort_values(by='horsepower',ascending = False).head(1)
#df.sort_values(by='horsepower',ascending = False).head(1)
```

```python

# 6. company wise average price of cars and order them in descending order
df.groupby('company')['price'].mean().round(2).sort_values(ascending=False)
```

```python
# 7. create a column named: "decision", If wheel-base is more than 100 and length is 
#     more than 175 then "Buy" else "Don't buy"?
df['decision'] = np.where( (df['wheel_base']>100)& (df['length']>175),'Buy',"Don't buy")
df.head(10)
```

```python

# 8. Select only cars whose decision is "buy"?
df[df['decision'] == 'Buy']

```

```python
# 9. Find is there any relation between horse power and mileage? 
df.plot(kind='scatter',x='horsepower',y='mileage')
plt.show()
```
